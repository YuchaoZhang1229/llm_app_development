## FastAPI 简介

**FastAPI 是什么**： 一个基于 Python 类型注解的现代 Web 框架，底层使用 `Starlette（异步 HTTP）` 和 `Pydantic（数据校验）`，自动生成 OpenAPI 文档。

**核心优势**：
- 高性能：基于 ASGI，支持异步，性能接近 Node.js / Go
- 类型安全：利用 Python 类型注解自动校验请求/响应数据
- 自动文档：内置 Swagger UI 和 ReDoc，零配置生成 API 文档
- 快速开发：声明式路由 + 依赖注入，减少样板代码

**学习前提**
- 熟悉 Python 基础知识（变量、数据类型、函数、类等）
- 了解异步编程概念(async/await)
- 掌握基本的 HTTP 协议知识

**理解 ASGI 与 WSGI 的区别**
- WSGI（Web Server Gateway Interface）：传统 Python Web 标准，同步处理，每个请求占用一个线程。Flask、Django 默认基于此。
- ASGI（Asynchronous Server Gateway Interface）：异步标准，支持 async/await、WebSocket、长连接。FastAPI 基于此。

简单理解：WSGI 像一条车道一次过一辆车，ASGI 像多车道并行，同一时间可以处理更多请求。在 I/O 密集型场景（数据库查询、调用外部 API）中，ASGI 的并发优势尤其明显。

## 第一个 FastAPI 应用

① 创建一个名为`main.py`的文件，内容如下：

````python
from fastapi import FastAPI

# 创建FastAPI应用实例
app = FastAPI()

# http://localhost:8000/
@app.get("/")
async def root():
    return {"message": "Hello, FastAPI!"}

# http://localhost:8000/items/100?q1=asjfda
@app.get("/items/{item_id}")
async def read_item(item_id: int, q1: str, q2: str | None = None):
    return {"item_id": item_id, "q1": q1, "q2": q2}
````

② 在终端中运行以下命令：
```bash
uvicorn main:app --reload    # 开发模式
uvicorn main:app --workers 4 # 生产模式
```
- `main`: 包含 FastAPI 应用的 Python 文件（不包含.py 后缀）。
- `app`: 在 main.py 文件中创建的 FastAPI 应用实例（app = FastAPI ()）。
- `--reload`: 开发模式下的自动重载选项，修改代码后自动重启服务器。


③ 访问 API

打开浏览器或使用工具（如 Postman、curl）访问：

- <http://localhost:8000/>

④ 查看自动生成的文档

FastAPI 自动生成交互式 API 文档：

- Swagger UI: <http://localhost:8000/docs>
- ReDoc: <http://localhost:8000/redoc>
