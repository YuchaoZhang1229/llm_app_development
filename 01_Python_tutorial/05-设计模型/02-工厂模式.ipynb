## 工厂模式（Factory Pattern）


核心作用：
- 当需要大量创建一个类的实例的时候，就可以使用工厂模式
- 把对象创建逻辑封装起来，调用者不直接写死类名

优点：
- 大批量创建对象的时候有统一的入口，易于代码维护
- 当发生修改，修改工厂类的创建方法即可
- 符合现实世界的模式，即由工厂来制作产品（对象）

适用场景：
- 多格式解析
- 支付网关切换
- 插件系统
- 不同策略/服务选择
- 对象创建逻辑复杂时

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount: float) -> str: ...

class Alipay(Payment):
    def pay(self, amount: float) -> str:
        return f"支付宝支付 {amount} 元"

class WechatPay(Payment):
    def pay(self, amount: float) -> str:
        return f"微信支付 {amount} 元"

def payment_factory(kind: str) -> Payment:
    if kind == "alipay":
        return Alipay()
    if kind == "wechat":
        return WechatPay()
    raise ValueError(f"未知支付方式: {kind}")

service = payment_factory("alipay")
print(service.pay(99.9))
```
